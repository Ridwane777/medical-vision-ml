### Conclusion

We have noticed irregularities in the file. Some lines contain two columns, while others have three. There are also instances of "0x000000" and some missing data. "0x000000" codes appear in place of some testimonials.

## Cleaning of the dataset

To do this, we alone will first replace the 0x000000 testimonials with the others. Then, remove the columns where testimonials are missing and any duplicates.

```python
df_clean = pd.read_csv('../Student_Dataset.csv', header=None, names=['ID', 'Code', 'Testimonies'], on_bad_lines='skip')

mask = df_clean['Testimonies'].isna() & (df_clean['Code'] != "0x000000")
df_clean.loc[mask, 'Testimonies'] = df_clean.loc[mask, 'Code']

df_final = df_clean[['ID', 'Testimonies']].copy()

df_final = df_final.dropna(subset=['Testimonies'])
df_final = df_final.drop_duplicates(subset=['Testimonies'])

print(f"Dimensions finales : {df_final.shape}")
print("-" * 30)

df_final.to_csv('Student_Dataset_Cleaned.csv', index=False)
df_final.head()
```
